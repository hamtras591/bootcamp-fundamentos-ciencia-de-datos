## ¿Qué es Matplotlib?

**Matplotlib** es la biblioteca más utilizada para crear visualizaciones en Python. Creada en 2003 por John D. Hunter, se ha convertido en el estándar para gráficos científicos y análisis de datos.

### ¿Por qué es importante?

- 📊 **Explorar datos**: Identificar patrones, tendencias y anomalías
- 📈 **Comunicar resultados**: Presentar hallazgos de forma clara
- 🔍 **Análisis visual**: Facilitar la comprensión de relaciones complejas
- 🎨 **Flexibilidad**: Desde gráficos simples hasta visualizaciones complejas
- 🔗 **Integración**: Funciona perfectamente con NumPy y Pandas

### Historia y contexto
```
2003: John D. Hunter crea Matplotlib
      └─ Inspirado en MATLAB

Hoy: Biblioteca estándar para visualización
     ├─ Ciencia
     ├─ Finanzas
     ├─ Ingeniería
     └─ Machine Learning
```

### Arquitectura básica
```
matplotlib
    └── pyplot (interfaz tipo MATLAB)
        ├── figure() - Crear figura
        ├── plot() - Gráfico de líneas
        ├── scatter() - Gráfico de dispersión
        ├── bar() - Gráfico de barras
        └── show() - Mostrar gráfico
```

## Gráfico de Líneas

Los **gráficos de líneas** son ideales para mostrar:
- 📈 Tendencias a lo largo del tiempo
- 🔄 Cambios continuos en los datos
- 📊 Comparaciones entre series temporales

### Anatomía de un gráfico de líneas
```
Título
    ↓
    ┌─────────────────────────────┐
    │         ╱                   │ ← Línea
Eje │      ╱                      │
Y   │   ╱                         │
    │╱                            │
    └─────────────────────────────┘
              Eje X
```

### Caso: Ventas mensuales de un producto

````python
# Datos de ejemplo: Ventas mensuales
meses = np.array(['Ene', 'Feb', 'Mar', 'Abr', 'May'])
ventas = np.array([20, 25, 30, 28, 35])

# Crear gráfico básico
plt.figure(figsize=(10, 6))
plt.plot(meses, ventas)
plt.title('Ventas mensuales - Versión básica')
plt.xlabel('Meses')
plt.ylabel('Ventas (miles de unidades)')
plt.show()
````

### Mejorando el gráfico: Agregar marcadores y color

````python
# Gráfico mejorado con marcadores y color
plt.figure(figsize=(10, 6))
plt.plot(meses, ventas,
         marker='s',           # Marcador circular
         color='blue',         # Color de línea
         linewidth=2,          # Grosor de línea
         markersize=4)         # Tamaño de marcador

plt.title('Ventas Mensuales de Producto', fontsize=16, fontweight='bold')
plt.xlabel('Meses', fontsize=12)
plt.ylabel('Ventas (en miles de unidades)', fontsize=12)
plt.grid(True, alpha=0.3)  # Agregar cuadrícula
plt.show()
````

### Tipos de marcadores disponibles

| Símbolo | Marcador | Símbolo | Marcador |
|---------|----------|---------|----------|
| `'o'` | Círculo | `'s'` | Cuadrado |
| `'v'` | Triángulo abajo | `'^'` | Triángulo arriba |
| `'*'` | Estrella | `'D'` | Diamante |
| `'+'` | Plus | `'x'` | X |

````python
# Ejemplo con diferentes marcadores
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.plot(meses, ventas, marker='o', label="Círculo")
plt.title('Marcador: Círculo')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(meses, ventas, marker='s', color='red', label="Cuadrado")
plt.title('Marcador: Cuadrado')
plt.legend()

plt.tight_layout()
plt.show()
````

### Múltiples líneas en un mismo gráfico

````python
# Datos para comparar dos productos
producto_a = np.array([20, 25, 30, 28, 35])
producto_b = np.array([15, 22, 28, 32, 30])

plt.figure(figsize=(10, 6))
plt.plot(meses, producto_a, marker='o', color='blue', linewidth=2, label='Producto A')
plt.plot(meses, producto_b, marker='s', color='red', linewidth=2, label='Producto B')

plt.title('Comparación de Ventas: Producto A vs B', fontsize=16, fontweight='bold')
plt.xlabel('Meses', fontsize=12)
plt.ylabel('Ventas (miles de unidades)', fontsize=12)
plt.legend(loc='best')  # Leyenda en mejor posición
plt.grid(True, alpha=0.3)
plt.show()
````

## Gráfico de Dispersión (Scatter Plot)

Los **gráficos de dispersión** son perfectos para:
- 🔗 Visualizar relaciones entre dos variables
- 📈 Identificar correlaciones (positivas, negativas, nulas)
- 🔍 Detectar outliers (valores atípicos)
- 📊 Analizar agrupaciones (clusters)

### Anatomía de un scatter plot
```
    Y
    ↑
  10│      •
    │    •   •
   5│  •       •
    │•           •
   0└──────────────→ X
     0  2  4  6  8
```

### Caso: Relación entre horas de estudio y puntaje en examen

````python
# Datos de ejemplo: Horas de estudio vs Puntaje
horas_estudio = np.array([1, 2, 3, 4, 5, 6, 7, 8])
puntaje_examen = np.array([55, 60, 65, 70, 75, 80, 85, 90])

# Gráfico de dispersión básico
plt.figure(figsize=(10, 6))
plt.scatter(horas_estudio, puntaje_examen)
plt.title('Horas de Estudio vs Puntaje - Versión básica')
plt.xlabel('Horas de Estudio')
plt.ylabel('Puntaje del Examen')
plt.show()
````

### Mejorando el scatter plot: Color, tamaño y transparencia

````python
# Scatter plot mejorado
plt.figure(figsize=(10, 6))
plt.scatter(horas_estudio, puntaje_examen,
           color='green',
           s=100,           # Tamaño de puntos (size)
           alpha=0.6,       # Transparencia
           edgecolors='black',  # Borde de puntos
           linewidth=1.5)

plt.title('Relación entre Horas de Estudio y Puntaje del Examen',
          fontsize=16, fontweight='bold')
plt.xlabel('Horas de Estudio', fontsize=12)
plt.ylabel('Puntaje del Examen (sobre 100)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
````

### Interpretación de correlaciones
```
Correlación Positiva      Correlación Negativa      Sin Correlación
       •                        •                        •
     •   •                    •   •                    •     •
   •       •                •       •                •   •     •
 -           •            •           •            •       •
```

- **Positiva**: ↗️ A mayor X, mayor Y
- **Negativa**: ↘️ A mayor X, menor Y
- **Nula**: No hay relación evidente

````python
# Ejemplo de diferentes correlaciones
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Correlación positiva
x_pos = np.array([1, 2, 3, 4, 5, 6, 7, 8])
y_pos = np.array([2, 4, 5, 7, 8, 10, 11, 13])
axes[0].scatter(x_pos, y_pos, color='green', s=100)
axes[0].set_title('Correlación Positiva', fontweight='bold')
axes[0].grid(True, alpha=0.3)

# Correlación negativa
y_neg = np.array([13, 11, 10, 8, 7, 5, 4, 2])
axes[1].scatter(x_pos, y_neg, color='red', s=100)
axes[1].set_title('Correlación Negativa', fontweight='bold')
axes[1].grid(True, alpha=0.3)

# Sin correlación
np.random.seed(42)
y_rand = np.random.randint(2, 14, size=8)
axes[2].scatter(x_pos, y_rand, color='gray', s=100)
axes[2].set_title('Sin Correlación', fontweight='bold')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## Personalización Básica

### Elementos clave de un gráfico
```
                 Título
                    ↓
    ┌─────────────────────────────┐
    │ Leyenda →  ─── Serie 1      │
Eje │             ─── Serie 2     │
Y   │         ╱                   │
    │      ╱                      │
    │   ╱                         │
    └─────────────────────────────┘
           Etiqueta Eje X
```

### Parámetros comunes de personalización

| Parámetro | Descripción | Valores |
|-----------|-------------|---------|
| `color` | Color de línea/puntos | 'blue', 'red', '#FF5733' |
| `linewidth` | Grosor de línea | 1, 2, 3, ... |
| `marker` | Tipo de marcador | 'o', 's', '^', '*' |
| `markersize` | Tamaño de marcador | 5, 10, 15, ... |
| `alpha` | Transparencia | 0.0 a 1.0 |
| `label` | Etiqueta para leyenda | 'Serie 1', 'Datos' |

````python
# Gráfico completamente personalizado
fig, ax = plt.subplots(figsize=(12, 7))

# Datos
x = np.linspace(0, 10, 50)
y1 = np.sin(x)
y2 = np.cos(x)

# Líneas personalizadas
ax.plot(x, y1, color='#FF6B6B', linewidth=2.5, linestyle='-',
        marker='o', markersize=6, label='Seno')
ax.plot(x, y2, color='#4ECDC4', linewidth=2.5, linestyle='--',
        marker='s', markersize=6, label='Coseno')

# Personalización
ax.set_title('Funciones Trigonométricas', fontsize=18, fontweight='bold', pad=20)
ax.set_xlabel('Ángulo (radianes)', fontsize=14)
ax.set_ylabel('Valor', fontsize=14)
ax.legend(loc='upper right', fontsize=12, frameon=True, shadow=True)
ax.grid(True, alpha=0.3, linestyle=':', linewidth=1)
ax.axhline(y=0, color='black', linewidth=0.8)  # Línea horizontal en y=0
ax.axvline(x=0, color='black', linewidth=0.8)  # Línea vertical en x=0

plt.tight_layout()
plt.show()
````

### Estilos de línea disponibles

| Código | Estilo |
|--------|--------|
| `'-'` | Línea continua |
| `'--'` | Línea discontinua |
| `'-.'` | Línea punto-raya |
| `':'` | Línea punteada |
| `''` | Sin línea (solo marcadores) |

````python
# Demostración de estilos de línea
fig, ax = plt.subplots(figsize=(12, 6))

x = np.linspace(0, 10, 20)
estilos = ['-', '--', '-.', ':']
etiquetas = ['Continua', 'Discontinua', 'Punto-raya', 'Punteada']

for i, (estilo, etiqueta) in enumerate(zip(estilos, etiquetas)):
    y = np.ones_like(x) * (i + 1)
    ax.plot(x, y, linestyle=estilo, linewidth=2, label=etiqueta)

ax.set_title('Estilos de Línea en Matplotlib', fontsize=16, fontweight='bold')
ax.set_xlabel('X', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.legend(loc='best', fontsize=11)
ax.set_ylim(0, 5)
ax.grid(True, alpha=0.3)
plt.show()
````

## Casos de Uso en Data Science

### Caso de Uso 1: Análisis de Tendencia de Ventas

**Contexto:** Eres analista de una empresa retail y necesitas presentar al equipo ejecutivo cómo han evolucionado las ventas en el último año para identificar estacionalidad y tomar decisiones de inventario.

**Objetivo:** Crear una visualización clara de ventas mensuales con indicadores clave.

````python
# Datos de ventas del último año
meses_completo = np.array(['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun',
                           'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic'])
ventas_2024 = np.array([120, 135, 150, 145, 160, 155,
                        140, 165, 180, 195, 220, 250])

# Calcular media de ventas
media_ventas = ventas_2024.mean()

print(f"📊 Análisis de Ventas 2024")
print(f"💰 Ventas totales: ${ventas_2024.sum()}K")
print(f"📈 Promedio mensual: ${media_ventas:.2f}K")
print(f"🏆 Mejor mes: {meses_completo[ventas_2024.argmax()]} (${ventas_2024.max()}K)")
print(f"📉 Peor mes: {meses_completo[ventas_2024.argmin()]} (${ventas_2024.min()}K)")
````

````python
# Visualización profesional para presentación ejecutiva
fig, ax = plt.subplots(figsize=(14, 7))

# Línea principal de ventas
ax.plot(meses_completo, ventas_2024,
        marker='o',
        color='#2E86AB',
        linewidth=3,
        markersize=10,
        label='Ventas Mensuales')

# Línea de promedio
ax.axhline(y=media_ventas,
           color='#E63946',
           linestyle='--',
           linewidth=2,
           label=f'Promedio: ${media_ventas:.0f}K')

# Resaltar mejor mes
mejor_mes_idx = ventas_2024.argmax()
ax.scatter(mejor_mes_idx, ventas_2024[mejor_mes_idx],
          color='#06FFA5', s=300, zorder=5,
          edgecolors='black', linewidth=2)

# Anotación del mejor mes
ax.annotate(f'Pico: ${ventas_2024.max()}K',
           xy=(mejor_mes_idx, ventas_2024[mejor_mes_idx]),
           xytext=(mejor_mes_idx-1, ventas_2024.max()+15),
           fontsize=12,
           fontweight='bold',
           arrowprops=dict(arrowstyle='->', color='black', lw=2))

# Personalización
ax.set_title('Evolución de Ventas 2024 - Dashboard Ejecutivo',
            fontsize=18, fontweight='bold', pad=20)
ax.set_xlabel('Mes', fontsize=14, fontweight='bold')
ax.set_ylabel('Ventas (miles de $)', fontsize=14, fontweight='bold')
ax.legend(loc='upper left', fontsize=12, frameon=True, shadow=True)
ax.grid(True, alpha=0.3, linestyle=':', linewidth=1)
ax.set_ylim(100, 270)

plt.tight_layout()
plt.show()
````

**🎯 Insights del Caso 1:**

✅ **Tendencia clara**: Crecimiento sostenido durante el año
✅ **Estacionalidad**: Pico en Nov-Dic (temporada navideña)
✅ **Caída leve**: Julio-Agosto (verano, baja demanda)
✅ **Crecimiento anual**: 108% entre Enero y Diciembre

**Decisiones de negocio:**
- 📦 Aumentar inventario en Q4 (Oct-Dic)
- 💡 Promociones en verano para impulsar ventas bajas
- 📈 Proyección 2025: Iniciar año con mayor stock base
- 🎯 Meta realista: Crecer 15-20% sobre 2024

### Caso de Uso 2: Análisis de Correlación - Inversión en Marketing vs Ingresos

**Contexto:** El equipo de marketing quiere demostrar que el presupuesto invertido en publicidad digital tiene impacto directo en los ingresos. Necesitas un gráfico que muestre esta relación para justificar aumento de presupuesto.

**Objetivo:** Visualizar correlación entre inversión y retorno.

````python
# Datos de inversión en marketing vs ingresos generados (en miles)
inversion_marketing = np.array([10, 15, 12, 20, 18, 25, 22, 30, 28, 35, 32, 40])
ingresos_generados = np.array([50, 65, 58, 85, 78, 105, 95, 125, 118, 145, 135, 165])

# Calcular ROI (Return on Investment)
roi = (ingresos_generados / inversion_marketing)

print("📊 Análisis de Marketing ROI")
print(f"💰 Inversión total: ${inversion_marketing.sum()}K")
print(f"💵 Ingresos generados: ${ingresos_generados.sum()}K")
print(f"📈 ROI promedio: {roi.mean():.2f}x")
print(f"🎯 Por cada $1 invertido, se generan ${roi.mean():.2f}")
````

````python
# Visualización de correlación con línea de tendencia
fig, ax = plt.subplots(figsize=(12, 7))

# Scatter plot
scatter = ax.scatter(inversion_marketing, ingresos_generados,
                    c=roi,  # Color por ROI
                    s=200,
                    cmap='RdYlGn',  # Colormap: rojo a verde
                    alpha=0.7,
                    edgecolors='black',
                    linewidth=2)

# Línea de tendencia (regresión lineal simple)
z = np.polyfit(inversion_marketing, ingresos_generados, 1)
p = np.poly1d(z)
ax.plot(inversion_marketing, p(inversion_marketing),
       "r--", linewidth=2, alpha=0.8, label='Tendencia')

# Barra de color para ROI
cbar = plt.colorbar(scatter, ax=ax)
cbar.set_label('ROI (X veces)', fontsize=12, fontweight='bold')

# Personalización
ax.set_title('Inversión en Marketing vs Ingresos Generados\n(Análisis de ROI)',
            fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Inversión en Marketing ($K)', fontsize=13, fontweight='bold')
ax.set_ylabel('Ingresos Generados ($K)', fontsize=13, fontweight='bold')
ax.legend(loc='upper left', fontsize=11)
ax.grid(True, alpha=0.3, linestyle=':', linewidth=1)

# Anotación
ax.text(0.05, 0.90,
       f'Correlación: Fuerte positiva\nROI Promedio: {roi.mean():.2f}x',
       transform=ax.transAxes,
       fontsize=11,
       verticalalignment='top',
       bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.show()
````

**🎯 Insights del Caso 2:**

✅ **Correlación positiva fuerte**: Más inversión = Más ingresos
✅ **ROI consistente**: Entre 4x y 5x de retorno
✅ **Tendencia lineal**: Relación predecible y escalable
✅ **Punto óptimo**: Inversiones de $30K+ generan mejor ROI

**Justificación para stakeholders:**
- 💰 Cada $1K adicional genera ~$4.1K en ingresos
- 📈 Modelo escalable: Se puede proyectar con confianza
- 🎯 Propuesta: Aumentar presupuesto 30% (de $40K a $52K)
- 💡 Retorno esperado: $52K × 4.1 = $213K de ingresos

**Aplicación al proyecto CSAT:**
Similar análisis correlacional puede mostrar si mayor satisfacción del cliente (CSAT) se correlaciona con mayores ingresos o retención.
