## Concepto

El **filtrado de datos** es una técnica fundamental que permite seleccionar subconjuntos específicos de un DataFrame basándose en condiciones lógicas.

### ¿Por qué filtrar datos?

- 🎯 **Enfoque**: Concentrarse en datos relevantes para el análisis
- 🔍 **Exploración**: Investigar segmentos específicos del dataset
- 📊 **Análisis segmentado**: Comparar comportamientos entre grupos
- 🚀 **Rendimiento**: Trabajar con subconjuntos más pequeños y rápidos
- 💡 **Insights**: Descubrir patrones en grupos específicos

### Anatomía de un filtro en Pandas
```python
# Sintaxis básica
df_filtrado = df[condicion]

# Ejemplo visual
df = df[df['columna'] > 100]
    └─┬─┘ └────────┬────────┘
      │           └─ Condición (retorna True/False)
      └─ DataFrame filtrado
```

### Tipos de filtrado

| Tipo | Descripción | Ejemplo |
|------|-------------|---------|
| **Simple** | Una sola condición | `df[df['País'] == 'Colombia']` |
| **Comparación** | Operadores >, <, >=, <= | `df[df['Ventas'] > 1000]` |
| **Múltiple (AND)** | Todas las condiciones deben cumplirse | `df[(cond1) & (cond2)]` |
| **Múltiple (OR)** | Al menos una condición debe cumplirse | `df[(cond1) | (cond2)]` |
| **Temporal** | Basado en fechas | `df[df['Fecha'].dt.year == 2024]` |
| **Negación** | Lo opuesto a la condición | `df[~(df['País'] == 'USA')]` |

````python
from data_loader import load_data
import pandas as pd
import numpy as np
````

````python
# Cargar datos de ejemplo
df = load_data('online_retail.csv')

# Convertir InvoiceDate a datetime
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

# Crear columna TotalPrice si no existe
if 'TotalPrice' not in df.columns:
    df['TotalPrice'] = df['Quantity'] * df['UnitPrice']

print("📦 Dataset original:")
print(f"Forma: {df.shape}")
print(f"\nPrimeras filas:")
print(df.head())
````

## Filtrado Básico por Condición

### Filtro de igualdad (==)

La forma más simple de filtrar es buscar coincidencias exactas en una columna.

**Sintaxis:**
```python
df[df['columna'] == 'valor']
```

````python
# Filtrar ventas del Reino Unido
uk_sales = df[df['Country'] == 'United Kingdom']

print("🇬🇧 Ventas en Reino Unido:")
print(f"Total de registros: {len(uk_sales):,}")
print(f"Porcentaje del total: {(len(uk_sales)/len(df)*100):.2f}%")
print(f"\nPrimeras filas:")
print(uk_sales.head())
````

### ⚠️ Cuidado con espacios y mayúsculas

Los filtros distinguen entre mayúsculas/minúsculas y espacios:

````python
# Demostrar diferencias
print("Valores únicos en la columna Country:")
print(df['Country'].unique()[:10])

# Filtro incorrecto (no encuentra nada)
filtro_incorrecto = df[df['Country'] == 'united kingdom']
print(f"\n❌ Filtro con minúsculas: {len(filtro_incorrecto)} registros")

# Filtro correcto
filtro_correcto = df[df['Country'] == 'United Kingdom']
print(f"✅ Filtro con mayúsculas: {len(filtro_correcto)} registros")
````

### Filtros con múltiples valores: isin()

Para filtrar por varios valores posibles, usa `isin()`:

````python
# Filtrar ventas de múltiples países
paises_europa = ['France', 'Germany', 'Spain', 'Italy', 'Netherlands']
ventas_europa = df[df['Country'].isin(paises_europa)]

print("🌍 Ventas en países europeos seleccionados:")
print(f"Total de registros: {len(ventas_europa):,}")
print("\n📊 Distribución por país:")
print(ventas_europa['Country'].value_counts())
````

## Operadores de Comparación

Pandas soporta todos los operadores de comparación estándar:

| Operador | Significado | Ejemplo |
|----------|-------------|---------|
| `==` | Igual a | `df[df['Cantidad'] == 5]` |
| `!=` | Diferente de | `df[df['País'] != 'USA']` |
| `>` | Mayor que | `df[df['Precio'] > 100]` |
| `>=` | Mayor o igual que | `df[df['Stock'] >= 50]` |
| `<` | Menor que | `df[df['Descuento'] < 0.2]` |
| `<=` | Menor o igual que | `df[df['Edad'] <= 30]` |

### Ejemplo: Mayor que (>)

````python
# Filtrar ventas con cantidad mayor a 10
high_quantity = df[df['Quantity'] > 10]

print("📦 Ventas con cantidad > 10:")
print(f"Total de registros: {len(high_quantity):,}")
print(f"Porcentaje del total: {(len(high_quantity)/len(df)*100):.2f}%")
print(f"\nEstadísticas de Quantity en ventas filtradas:")
print(high_quantity['Quantity'].describe())
````

### Ejemplo: Menor que (<)

````python
# Filtrar productos con precio unitario bajo (< $2)
low_price = df[df['UnitPrice'] < 2]

print("💵 Productos con precio < $2:")
print(f"Total de registros: {len(low_price):,}")
print(f"\nTop 10 productos más baratos:")
print(low_price.nsmallest(10, 'UnitPrice')[['Description', 'UnitPrice']])
````

### Ejemplo: Diferente de (!=)

````python
# Filtrar ventas que NO son del Reino Unido
non_uk_sales = df[df['Country'] != 'United Kingdom']

print("🌍 Ventas fuera del Reino Unido:")
print(f"Total de registros: {len(non_uk_sales):,}")
print(f"\n📊 Top 10 países (excluyendo UK):")
print(non_uk_sales['Country'].value_counts().head(10))
````

## Filtrado con Múltiples Condiciones

Para combinar condiciones, usamos operadores lógicos:

| Operador | Significado | Uso |
|----------|-------------|-----|
| `&` | AND (Y) | Ambas condiciones deben ser True |
| `|` | OR (O) | Al menos una condición debe ser True |
| `~` | NOT (NO) | Niega la condición |

⚠️ **IMPORTANTE:** Siempre usa **paréntesis** alrededor de cada condición.

### Operador AND (&)

Todas las condiciones deben cumplirse simultáneamente.

````python
# Filtrar ventas del Reino Unido con cantidad > 40
high_uk_sales = df[
    (df['Country'] == 'United Kingdom') &
    (df['Quantity'] > 40)
]

print("🇬🇧📦 Ventas en UK con cantidad > 40:")
print(f"Total de registros: {len(high_uk_sales):,}")
print(f"\nEstadísticas:")
print(high_uk_sales[['Quantity', 'UnitPrice', 'TotalPrice']].describe())
````

### Ejemplo: Tres condiciones con AND

````python
# Filtrar ventas premium: UK, cantidad alta, precio alto
premium_sales = df[
    (df['Country'] == 'United Kingdom') &
    (df['Quantity'] > 20) &
    (df['UnitPrice'] > 5)
]

print("💎 Ventas premium (UK, Cantidad > 20, Precio > $5):")
print(f"Total de registros: {len(premium_sales):,}")
print(f"Ingreso total: ${premium_sales['TotalPrice'].sum():,.2f}")
print(f"\nTop 10 productos premium:")
print(premium_sales.groupby('Description')['TotalPrice'].sum().nlargest(10))
````

### Operador OR (|)

Al menos una de las condiciones debe cumplirse.

````python
# Filtrar ventas de Francia O Alemania
france_or_germany = df[
    (df['Country'] == 'France') |
    (df['Country'] == 'Germany')
]

print("🇫🇷🇩🇪 Ventas en Francia O Alemania:")
print(f"Total de registros: {len(france_or_germany):,}")
print(f"\n📊 Distribución:")
print(france_or_germany['Country'].value_counts())
````

### Operador NOT (~)

Niega una condición (lo opuesto).

````python
# Filtrar ventas que NO son cantidades pequeñas (> 5)
not_small_quantity = df[~(df['Quantity'] <= 5)]

print("📦 Ventas que NO son cantidades pequeñas:")
print(f"Total de registros: {len(not_small_quantity):,}")
print(f"Porcentaje del total: {(len(not_small_quantity)/len(df)*100):.2f}%")

# Equivalente más legible
same_result = df[df['Quantity'] > 5]
print(f"\n✅ Verificación: {len(not_small_quantity) == len(same_result)}")
````

### Combinación compleja: AND + OR

````python
# Ventas premium: (UK O France) Y cantidad > 30
complex_filter = df[
    ((df['Country'] == 'United Kingdom') | (df['Country'] == 'France')) &
    (df['Quantity'] > 30)
]

print("🎯 Ventas complejas (UK o France) con cantidad > 30:")
print(f"Total de registros: {len(complex_filter):,}")
print(f"\n📊 Por país:")
print(complex_filter['Country'].value_counts())
````

## Filtrado de Series Temporales

Cuando trabajamos con columnas de tipo `datetime`, podemos filtrar por componentes de fecha.

### Propiedades de datetime disponibles

| Propiedad | Descripción | Ejemplo |
|-----------|-------------|---------|
| `.dt.year` | Año | `df['Fecha'].dt.year == 2024` |
| `.dt.month` | Mes (1-12) | `df['Fecha'].dt.month == 12` |
| `.dt.day` | Día del mes | `df['Fecha'].dt.day == 15` |
| `.dt.dayofweek` | Día de la semana (0=lunes) | `df['Fecha'].dt.dayofweek == 5` |
| `.dt.quarter` | Trimestre (1-4) | `df['Fecha'].dt.quarter == 4` |
| `.dt.hour` | Hora (0-23) | `df['Fecha'].dt.hour >= 14` |

### Filtrar por año

````python
# Verificar que InvoiceDate es datetime
print(f"Tipo de InvoiceDate: {df['InvoiceDate'].dtype}")

# Filtrar ventas del año 2011
sales_2011 = df[df['InvoiceDate'].dt.year == 2011]

print("\n📅 Ventas del año 2011:")
print(f"Total de registros: {len(sales_2011):,}")
print(f"Rango de fechas: {sales_2011['InvoiceDate'].min()} a {sales_2011['InvoiceDate'].max()}")
print(f"Ingreso total 2011: ${sales_2011['TotalPrice'].sum():,.2f}")
````

### Filtrar por año y mes específico

````python
# Filtrar ventas de diciembre de 2010
december_2010 = df[
    (df['InvoiceDate'].dt.year == 2010) &
    (df['InvoiceDate'].dt.month == 12)
]

print("🎄 Ventas de Diciembre 2010:")
print(f"Total de registros: {len(december_2010):,}")
print(f"Ingreso total: ${december_2010['TotalPrice'].sum():,.2f}")
print(f"\n📊 Ventas por día:")
print(december_2010.groupby(december_2010['InvoiceDate'].dt.day)['TotalPrice'].sum())
print(f"Rango de fechas: {december_2010['InvoiceDate'].min()} a {december_2010['InvoiceDate'].max()}")
````

### Filtrar por día de la semana

````python
# Extraer día de la semana (0 = Lunes, 6 = Domingo)
df['DayOfWeek'] = df['InvoiceDate'].dt.dayofweek
df['DayName'] = df['InvoiceDate'].dt.day_name()

# Filtrar solo ventas de fin de semana (sábado y domingo)
weekend_sales = df[df['DayOfWeek'].isin([5, 6])]

print("🏖️ Ventas de fin de semana:")
print(f"Total de registros: {len(weekend_sales):,}")
print(f"Porcentaje del total: {(len(weekend_sales)/len(df)*100):.2f}%")
print(f"\n📊 Distribución por día:")
print(weekend_sales['DayName'].value_counts())
````

### Filtrar por rango de fechas

````python
# Definir rango de fechas
fecha_inicio = pd.to_datetime('2010-12-01')
fecha_fin = pd.to_datetime('2010-12-31')

# Filtrar ventas en el rango
ventas_rango = df[
    (df['InvoiceDate'] >= fecha_inicio) &
    (df['InvoiceDate'] <= fecha_fin)
]

print(f"📅 Ventas entre {fecha_inicio.date()} y {fecha_fin.date()}:")
print(f"Total de registros: {len(ventas_rango):,}")
print(f"Ingreso total: ${ventas_rango['TotalPrice'].sum():,.2f}")
print(f"Ticket promedio: ${ventas_rango['TotalPrice'].mean():.2f}")
````

## Filtrado Avanzado

### Método query() - Sintaxis más legible

El método `query()` permite escribir filtros con sintaxis SQL-like:

````python
# Usando query() en lugar de corchetes
uk_high_sales_query = df.query("Country == 'United Kingdom' and Quantity > 40")

print("🔍 Usando query():")
print(f"Total de registros: {len(uk_high_sales_query):,}")

# Comparar con método tradicional
uk_high_sales_traditional = df[
    (df['Country'] == 'United Kingdom') & (df['Quantity'] > 40)
]

print(f"\n✅ ¿Dan el mismo resultado? {len(uk_high_sales_query) == len(uk_high_sales_traditional)}")
````

### Ventajas de query()

- ✅ Más legible para filtros complejos
- ✅ No necesita paréntesis externos
- ✅ Usa `and`, `or` en lugar de `&`, `|`
- ✅ Soporta variables con `@`

**Ejemplo con variables:**

````python
# Usar variables en query()
pais = 'Germany'
cantidad_minima = 20

ventas_query = df.query("Country == @pais and Quantity > @cantidad_minima")

print(f"🇩🇪 Ventas en {pais} con cantidad > {cantidad_minima}:")
print(f"Total de registros: {len(ventas_query):,}")
````

### Filtrar con str.contains() - Búsqueda de texto

````python
# Buscar productos que contengan la palabra "HEART"
heart_products = df[df['Description'].str.contains('HEART', na=False, case=False)]

print("❤️ Productos con 'HEART' en descripción:")
print(f"Total de registros: {len(heart_products):,}")
print(f"\nProductos únicos:")
print(heart_products['Description'].unique()[:10])
````

### Filtrar con between() - Rangos numéricos

````python
# Filtrar precios en un rango específico
medium_price = df[df['UnitPrice'].between(5, 20)]

print("💰 Productos con precio entre $5 y $20:")
print(f"Total de registros: {len(medium_price):,}")
print(f"\nEstadísticas de precio:")
print(medium_price['UnitPrice'].describe())
````

## Casos de Uso en Data Science

### Caso de Uso 1: Análisis de Ventas Estacionales

**Contexto:** Eres analista de una tienda de retail. El equipo de marketing quiere lanzar campañas específicas basadas en patrones de compra estacionales y necesita identificar qué meses tienen mejor rendimiento y qué productos se venden más en temporadas específicas.

**Objetivo:** Analizar ventas por trimestre y mes para identificar patrones estacionales y productos estrella por temporada.

````python
# Preparar datos: agregar columnas temporales
df['Quarter'] = df['InvoiceDate'].dt.quarter
df['Month'] = df['InvoiceDate'].dt.month
df['MonthName'] = df['InvoiceDate'].dt.month_name()

print("📅 Datos preparados con información temporal")
print(df[['InvoiceDate', 'Quarter', 'Month', 'MonthName']].head())
````

**Paso 1: Análisis de ventas por trimestre**

````python
# Ventas por trimestre
quarterly_sales = df.groupby('Quarter').agg({
    'TotalPrice': ['sum', 'mean', 'count'],
    'Quantity': 'sum'
}).round(2)

quarterly_sales.columns = ['Revenue', 'Avg_Ticket', 'Num_Transactions', 'Total_Items']
quarterly_sales = quarterly_sales.reset_index()

print("📊 Ventas por trimestre:")
print(quarterly_sales)

# Identificar el mejor trimestre
best_quarter = quarterly_sales.loc[quarterly_sales['Revenue'].idxmax(), 'Quarter']
print(f"\n🏆 Mejor trimestre: Q{best_quarter}")
````

**Paso 2: Análisis detallado del Q4 (temporada alta)**

````python
# Filtrar ventas del Q4 (octubre-diciembre)
q4_sales = df[df['Quarter'] == 4]

print(f"🎄 Análisis del Q4 (Temporada Alta):")
print(f"Total de transacciones: {len(q4_sales):,}")
print(f"Ingreso total: ${q4_sales['TotalPrice'].sum():,.2f}")

# Ventas por mes en Q4
q4_monthly = q4_sales.groupby('MonthName')['TotalPrice'].agg(['sum', 'count'])
q4_monthly.columns = ['Revenue', 'Transactions']
print(f"\n📈 Ventas mensuales en Q4:")
print(q4_monthly)
print(q4_monthly.info())
````

**Paso 3: Top productos por temporada**

````python
# Top 10 productos en Q4
top_q4_products = q4_sales.groupby('Description')['TotalPrice'].sum().nlargest(10)

print("🌟 Top 10 productos en Q4:")
for i, (product, revenue) in enumerate(top_q4_products.items(), 1):
    print(f"{i}. {product[:50]}: ${revenue:,.2f}")

# Comparar con Q2 (verano)
q2_sales = df[df['Quarter'] == 2]
top_q2_products = q2_sales.groupby('Description')['TotalPrice'].sum().nlargest(10)

print("\n☀️ Top 10 productos en Q2:")
for i, (product, revenue) in enumerate(top_q2_products.items(), 1):
    print(f"{i}. {product[:50]}: ${revenue:,.2f}")
````

**Paso 4: Identificar productos estacionales vs perennes**

````python
# Productos que aparecen en top 10 de todos los trimestres (perennes)
top_products_by_q = {}
for q in [1, 2, 3, 4]:
    q_data = df[df['Quarter'] == q]
    top_products_by_q[q] = set(q_data.groupby('Description')['TotalPrice'].sum().nlargest(10).index)

# Productos perennes (en todos los trimestres)
perennial_products = top_products_by_q[1] & top_products_by_q[2] & top_products_by_q[3] & top_products_by_q[4]
print(f"🌟 Productos perennes (top en todos los trimestres): {len(perennial_products)}")
for product in perennial_products:
    print(f"  - {product}")

# Productos exclusivos de Q4
q4_exclusive = top_products_by_q[4] - (top_products_by_q[1] | top_products_by_q[2] | top_products_by_q[3])
print(f"\n🎄 Productos exclusivos de Q4: {len(q4_exclusive)}")
for product in q4_exclusive:
    print(f"  - {product}")
````

**🎯 Insights del Caso 1:**

✅ **Patrón estacional claro**: Q4 tiene 40-50% más ingresos que otros trimestres
✅ **Noviembre = pico**: El mes con mayor actividad del año
✅ **Productos estacionales identificados**: Items navideños dominan Q4
✅ **Productos perennes**: ~3-5 productos se venden bien todo el año

**Recomendaciones estratégicas:**
1. 📈 **Stock**: Aumentar inventario de productos Q4 desde septiembre
2. 🎯 **Marketing**: Campañas navideñas desde inicio de noviembre
3. 💎 **Productos perennes**: Mantener stock constante todo el año
4. 🌱 **Q2 y Q3**: Desarrollar promociones para impulsar ventas en trimestres bajos

### Caso de Uso 2: Segmentación de Clientes por Comportamiento de Compra

**Contexto:** El equipo de CRM necesita segmentar clientes para personalizar comunicaciones. Quieren identificar diferentes tipos de compradores basándose en frecuencia, valor y recencia de compras usando filtros avanzados.

**Objetivo:** Crear segmentos accionables de clientes usando filtros complejos y análisis de comportamiento.

````python
# Preparar datos de clientes
customers_df = df[df['CustomerID'].notna()].copy()

# Fecha de análisis
analysis_date = customers_df['InvoiceDate'].max()

print(f"👥 Datos de clientes:")
print(f"Total transacciones con CustomerID: {len(customers_df):,}")
print(f"Clientes únicos: {customers_df['CustomerID'].nunique():,}")
print(f"Fecha de análisis: {analysis_date}")
````

**Paso 1: Calcular métricas por cliente**

````python
# Calcular métricas por cliente
customer_metrics = customers_df.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (analysis_date - x.max()).days,  # Recency
    'InvoiceNo': 'nunique',  # Frequency
    'TotalPrice': 'sum'  # Monetary
}).reset_index()

customer_metrics.columns = ['CustomerID', 'Days_Since_Last_Purchase', 'Num_Purchases', 'Total_Spent']

# Calcular ticket promedio
customer_metrics['Avg_Ticket'] = (customer_metrics['Total_Spent'] /
                                   customer_metrics['Num_Purchases']).round(2)

print("📊 Métricas por cliente:")
print(customer_metrics.head(10))
print(f"\nEstadísticas generales:")
print(customer_metrics.describe())
````

**Paso 2: Filtrar clientes VIP (alta frecuencia + alto valor)**

````python
# Definir umbrales para VIP
frequency_threshold = customer_metrics['Num_Purchases'].quantile(0.75)  # Top 25%
monetary_threshold = customer_metrics['Total_Spent'].quantile(0.75)     # Top 25%

# Filtrar clientes VIP
vip_customers = customer_metrics[
    (customer_metrics['Num_Purchases'] >= frequency_threshold) &
    (customer_metrics['Total_Spent'] >= monetary_threshold)
]

print("💎 Clientes VIP:")
print(f"Total: {len(vip_customers):,} clientes ({len(vip_customers)/len(customer_metrics)*100:.1f}%)")
print(f"Ingresos totales VIP: ${vip_customers['Total_Spent'].sum():,.2f}")
print(f"Porcentaje de ingresos: {vip_customers['Total_Spent'].sum()/customer_metrics['Total_Spent'].sum()*100:.1f}%")
print(f"\nEstadísticas VIP:")
print(vip_customers[['Num_Purchases', 'Total_Spent', 'Avg_Ticket']].describe())
````

**Paso 3: Filtrar clientes en riesgo (buenos históricamente pero inactivos)**

````python
# Clientes en riesgo: alto valor histórico pero no compran recientemente
at_risk_customers = customer_metrics[
    (customer_metrics['Total_Spent'] >= monetary_threshold) &
    (customer_metrics['Days_Since_Last_Purchase'] > 90)
]

print("⚠️ Clientes en riesgo:")
print(f"Total: {len(at_risk_customers):,} clientes")
print(f"Ingresos históricos en riesgo: ${at_risk_customers['Total_Spent'].sum():,.2f}")
print(f"Días promedio sin comprar: {at_risk_customers['Days_Since_Last_Purchase'].mean():.0f}")
print(f"\nTop 10 clientes en riesgo por valor:")
print(at_risk_customers.nlargest(10, 'Total_Spent')[['CustomerID', 'Total_Spent', 'Days_Since_Last_Purchase', 'Num_Purchases']])
````

**Paso 4: Filtrar nuevos clientes prometedores (baja frecuencia pero alta primera compra)**

````python
# Nuevos clientes prometedores: 1-2 compras pero ticket alto
promising_customers = customer_metrics[
    (customer_metrics['Num_Purchases'] <= 2) &
    (customer_metrics['Avg_Ticket'] > customer_metrics['Avg_Ticket'].median()) &
    (customer_metrics['Days_Since_Last_Purchase'] <= 30)
]

print("🌱 Clientes prometedores (nuevos con alto potencial):")
print(f"Total: {len(promising_customers):,} clientes")
print(f"Ticket promedio: ${promising_customers['Avg_Ticket'].mean():.2f}")
print(f"Días promedio desde última compra: {promising_customers['Days_Since_Last_Purchase'].mean():.0f}")
print(f"\nPotencial de ingresos si mantienen comportamiento:")
potential_revenue = promising_customers['Avg_Ticket'].sum() * 10  # Si compran 10 veces más
print(f"${potential_revenue:,.2f}")
````

**Paso 5: Filtrar clientes perdidos (no compran hace mucho tiempo)**

````python
# Clientes perdidos: más de 180 días sin comprar
lost_customers = customer_metrics[
    customer_metrics['Days_Since_Last_Purchase'] > 180
]

print("🚨 Clientes perdidos (>180 días sin comprar):")
print(f"Total: {len(lost_customers):,} clientes ({len(lost_customers)/len(customer_metrics)*100:.1f}%)")
print(f"Ingresos históricos perdidos: ${lost_customers['Total_Spent'].sum():,.2f}")
print(f"\nDistribución por tiempo sin comprar:")
print(pd.cut(lost_customers['Days_Since_Last_Purchase'],
             bins=[180, 270, 365, 500],
             labels=['6-9 meses', '9-12 meses', '>12 meses']).value_counts())
````

**Paso 6: Crear resumen completo de segmentación**

````python
# Crear función de segmentación
def segment_customer(row):
    """Clasifica cliente según métricas RFM."""
    days = row['Days_Since_Last_Purchase']
    purchases = row['Num_Purchases']
    spent = row['Total_Spent']

    # Umbrales
    freq_high = customer_metrics['Num_Purchases'].quantile(0.75)
    money_high = customer_metrics['Total_Spent'].quantile(0.75)

    # Segmentación
    if purchases >= freq_high and spent >= money_high and days <= 90:
        return 'VIP Active'
    elif spent >= money_high and days > 90:
        return 'At Risk'
    elif purchases <= 2 and row['Avg_Ticket'] > customer_metrics['Avg_Ticket'].median() and days <= 30:
        return 'Promising'
    elif days > 180:
        return 'Lost'
    elif purchases >= freq_high:
        return 'Loyal'
    elif spent >= money_high:
        return 'Big Spender'
    else:
        return 'Regular'

# Aplicar segmentación
customer_metrics['Segment'] = customer_metrics.apply(segment_customer, axis=1)

# Resumen por segmento
segment_summary = customer_metrics.groupby('Segment').agg({
    'CustomerID': 'count',
    'Total_Spent': ['sum', 'mean'],
    'Num_Purchases': 'mean',
    'Days_Since_Last_Purchase': 'mean'
}).round(2)

segment_summary.columns = ['Num_Customers', 'Total_Revenue', 'Avg_Revenue', 'Avg_Purchases', 'Avg_Days_Inactive']
segment_summary = segment_summary.reset_index()

# Calcular % de ingresos
segment_summary['Revenue_Pct'] = (segment_summary['Total_Revenue'] /
                                   segment_summary['Total_Revenue'].sum() * 100).round(2)

print("📊 Resumen de segmentación:")
print(segment_summary.sort_values('Total_Revenue', ascending=False))
````

**Paso 7: Estrategias específicas por segmento**

````python
# Definir estrategias por segmento
strategies = {
    'VIP Active': {
        'emoji': '💎',
        'action': 'Programa de fidelidad exclusivo',
        'campaign': 'Early access a nuevos productos',
        'budget': 'Alto',
        'priority': 1
    },
    'At Risk': {
        'emoji': '⚠️',
        'action': 'Campaña de recuperación urgente',
        'campaign': 'Descuento 20% + envío gratis',
        'budget': 'Alto',
        'priority': 2
    },
    'Promising': {
        'emoji': '🌱',
        'action': 'Onboarding mejorado',
        'campaign': 'Descuento en segunda compra',
        'budget': 'Medio',
        'priority': 3
    },
    'Loyal': {
        'emoji': '💙',
        'action': 'Cross-selling',
        'campaign': 'Recomendaciones personalizadas',
        'budget': 'Medio',
        'priority': 4
    },
    'Big Spender': {
        'emoji': '💰',
        'action': 'Productos premium',
        'campaign': 'Colecciones exclusivas',
        'budget': 'Medio',
        'priority': 5
    },
    'Lost': {
        'emoji': '🚨',
        'action': 'Reactivación agresiva',
        'campaign': 'Descuento 30% "Te extrañamos"',
        'budget': 'Bajo',
        'priority': 6
    },
    'Regular': {
        'emoji': '👤',
        'action': 'Comunicación regular',
        'campaign': 'Newsletter semanal',
        'budget': 'Bajo',
        'priority': 7
    }
}

# Mostrar estrategias con métricas
print("🎯 Plan de acción por segmento:\n")
for segment in segment_summary.sort_values('Revenue_Pct', ascending=False)['Segment']:
    segment_data = segment_summary[segment_summary['Segment'] == segment].iloc[0]
    strategy = strategies.get(segment, {})

    print(f"{strategy.get('emoji', '📌')} {segment}")
    print(f"   Clientes: {segment_data['Num_Customers']:.0f} ({segment_data['Num_Customers']/len(customer_metrics)*100:.1f}%)")
    print(f"   Ingresos: ${segment_data['Total_Revenue']:,.2f} ({segment_data['Revenue_Pct']}%)")
    print(f"   Estrategia: {strategy.get('action', 'N/A')}")
    print(f"   Campaña: {strategy.get('campaign', 'N/A')}")
    print(f"   Presupuesto: {strategy.get('budget', 'N/A')}")
    print(f"   Prioridad: {strategy.get('priority', 'N/A')}")
    print()
````

### Cheat Sheet de Operadores

| Operador | Uso | Ejemplo |
|----------|-----|---------|
| `==` | Igualdad | `df[df['col'] == 'valor']` |
| `!=` | Desigualdad | `df[df['col'] != 'valor']` |
| `>` | Mayor que | `df[df['col'] > 10]` |
| `>=` | Mayor o igual | `df[df['col'] >= 10]` |
| `<` | Menor que | `df[df['col'] < 10]` |
| `<=` | Menor o igual | `df[df['col'] <= 10]` |
| `&` | AND lógico | `df[(cond1) & (cond2)]` |
| `|` | OR lógico | `df[(cond1) | (cond2)]` |
| `~` | NOT lógico | `df[~(condicion)]` |
| `.isin([])` | Está en lista | `df[df['col'].isin(['A', 'B'])]` |
| `.between()` | En rango | `df[df['col'].between(5, 10)]` |
| `.str.contains()` | Contiene texto | `df[df['col'].str.contains('palabra')]` |

### Consejos para filtrado eficiente

1. **Usa query() para filtros complejos** - Más legible
2. **Siempre usa paréntesis** con `&` y `|`
3. **Verifica tipos de datos** antes de filtrar
4. **Maneja valores nulos** con `na=False` en string operations
5. **Combina con loc/iloc** para seleccionar columnas específicas
6. **Usa copy()** si modificarás el DataFrame filtrado
```python
# ✅ Buena práctica
filtrado = df[df['col'] > 10].copy()
filtrado['nueva_col'] = 100

# ❌ Evitar (SettingWithCopyWarning)
filtrado = df[df['col'] > 10]
filtrado['nueva_col'] = 100  # Advertencia
```
